```python
df.groupby("Pclass")["Survived"].mean()
```

```python
df.groupby("Pclass")["Survived"].mean().plot(kind="bar")

plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.title("Survival Rate by Passenger Class")

plt.show()
```

```python
df.groupby("Sex")["Survived"].mean()
```

```python
df.groupby("Sex")["Survived"].mean().plot(kind="bar")

plt.xlabel("Gender")
plt.ylabel("Survival Rate")
plt.title("Survival Rate by Gender")

plt.show()
```

```python
plt.scatter(df["Age"], df["Survived"])

plt.xlabel("Age")
plt.ylabel("Survived (0 = No, 1 = Yes)")
plt.title("Age vs Survival")

plt.show()
```

```python
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 12, 18, 30, 50, 80],
    labels=["Child", "Teen", "Young Adult", "Adult", "Senior"]
)
```

```python
df.groupby("AgeGroup", observed=True)["Survived"].mean() * 100
```

```python
df.groupby("AgeGroup", observed=True)["Survived"].mean().mul(100).plot(kind="bar")

plt.xlabel("Age Group")
plt.ylabel("Survival Rate (%)")
plt.title("Survival Rate by Age Group")

plt.show()
```

```python
df.groupby(["Sex", "Pclass"])["Survived"].mean() * 100
```

```python
survival_by_gender_class = (
    df.groupby(["Sex", "Pclass"])["Survived"]
    .mean()
    .unstack()
    * 100
)

print(survival_by_gender_class)
```

```python
survival_by_gender_class.plot(kind="bar")

plt.xlabel("Gender")
plt.ylabel("Survival Rate (%)")
plt.title("Survival Rate by Gender and Passenger Class")

plt.legend(title="Passenger Class")
plt.show()
```
